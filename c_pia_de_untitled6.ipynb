```python
import urllib.request
from bs4 import BeautifulSoup

url = "https://pt.wikipedia.org/wiki/Lista_de_animes_exibidos_no_Brasil"

try:
    # Use urllib.request para abrir a URL
    with urllib.request.urlopen(url) as response:
        html_content = response.read()

    # Use BeautifulSoup para parsear o conteúdo HTML
    site = BeautifulSoup(html_content, 'html.parser')

    print("Conteúdo da página obtido e parseado com sucesso!")
    # Agora você pode trabalhar com o objeto soup
    # Por exemplo, para imprimir o título da página:
    # print(soup.title)

except Exception as e:
    print(f"Ocorreu um erro ao tentar obter ou parsear a página: {e}")
```

## Extract data

### Subtask:

**Reasoning**:
Find all book elements, iterate through them, and extract the title, price, and rating into a list of dictionaries.



```python
anime_data = []

# Find all tables in the HTML. We'll assume the main list is in a table with class 'wikitable'.
tables = site.find_all('table', class_='wikitable')

# Iterate through the found tables
for table in tables:
    # Find all rows within the table body
    rows = table.find_all('tr')

    # Find the header row to get column names
    header_row = rows[0]
    headers = [header.get_text(strip=True) for header in header_row.find_all(['th', 'td'])]

    # Iterate through data rows (skip header row)
    for row in rows[1:]:
        cols = row.find_all('td')
        # Ensure the number of columns in the data row matches the number of headers
        if len(cols) == len(headers):
            row_data = {}
            for i, col in enumerate(cols):
                # Use the header text as the key for the data
                header = headers[i]
                row_data[header] = col.get_text(strip=True)
            anime_data.append(row_data)
        # Handle cases where a row might be a header within the table body or have a different structure
        elif row.find_all('th'):
             # This row is likely a sub-header, skip or handle differently if needed
             pass
        else:
             # Row has a different number of columns than headers, might need closer inspection
             # print(f"Skipping row with unexpected number of columns: {[col.get_text(strip=True) for col in cols]}")
             pass


# Display the first few extracted anime data
print(anime_data[:5])
print(f"Total rows extracted: {len(anime_data)}")
```
